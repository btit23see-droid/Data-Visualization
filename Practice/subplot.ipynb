Plot on Individual Axes

With plt.subplots(2, 2), we get four Axes.

We access them using indexing:

axes[0, 0] → top-left
axes[0, 1] → top-right
axes[1, 0] → bottom-left
axes[1, 1] → bottom-right

````python
# Create 2 × 2 subplot grid
fig, axes = plt.subplots(2, 2)

# Top-left subplot
axes[0, 0].plot([1, 2, 3], [2, 4, 6])
axes[0, 0].set_title("Plot 1")

# Top-right subplot
axes[0, 1].plot([1, 2, 3], [3, 6, 9])
axes[0, 1].set_title("Plot 2")

# Bottom-left subplot
axes[1, 0].plot([1, 2, 3], [4, 8, 12])
axes[1, 0].set_title("Plot 3")

# Bottom-right subplot
axes[1, 1].plot([1, 2, 3], [5, 10, 15])
axes[1, 1].set_title("Plot 4")

# Adjust spacing
plt.tight_layout()

# Display
plt.show()
````

````python
# Create 2 × 2 subplot grid
fig, axes = plt.subplots(2, 2)

# -------------------------
# Plot 1
# -------------------------
axes[0, 0].plot([1, 2, 3], [10, 20, 30])
axes[0, 0].set_title("Marks")
axes[0, 0].set_xlabel("Study Hours")
axes[0, 0].set_ylabel("Marks")


# -------------------------
# Plot 2
# -------------------------
axes[0, 1].plot([1, 2, 3], [100, 200, 300])
axes[0, 1].set_title("Sales")
axes[0, 1].set_xlabel("Months")
axes[0, 1].set_ylabel("Sales")


# -------------------------
# Plot 3
# -------------------------
axes[1, 0].plot([1, 2, 3], [25, 30, 35])
axes[1, 0].set_title("Temperature")
axes[1, 0].set_xlabel("Days")
axes[1, 0].set_ylabel("Temperature")


# -------------------------
# Plot 4
# -------------------------
axes[1, 1].plot([1, 2, 3], [50, 60, 70])
axes[1, 1].set_title("Population")
axes[1, 1].set_xlabel("Years")
axes[1, 1].set_ylabel("Population")


# Adjust layout
plt.tight_layout()

# Display
plt.show()
````

## Difference Between `plt.subplot()` and `plt.subplots()`

| `plt.subplot()` | `plt.subplots()` |
|---|---|
| Selects one subplot at a time | Creates the Figure and Axes together |
| Uses `plt.subplot(rows, columns, position)` | Uses `fig, axes = plt.subplots(rows, columns)` |
| We usually use `plt.plot()` | We use `axes.plot()` |
| Works with the **current Axes** | Gives us direct access to each **Axes object** |
| Example: `plt.subplot(2, 2, 1)` | Example: `fig, axes = plt.subplots(2, 2)` |
| Good for learning the basic subplot concept | Better for managing multiple plots |
| Convenient for simple plots | More flexible for complex plots |

### Example

#### Using `plt.subplot()`

```python
plt.subplot(2, 2, 1)
plt.plot(x, y)
```

#### Using `plt.subplots()`

```python
fig, axes = plt.subplots(2, 2)
axes[0, 0].plot(x, y)
```

Both draw the same line in the same top-left position — `plt.subplots()` just hands you
the `axes` object directly instead of making you re-select it with `plt.subplot()` each time.


## Controlling Figure Size with `figsize`

So far every figure has used Matplotlib's default size. `figsize=(width, height)` (in inches)
lets us make the whole figure bigger or smaller — useful when subplots start looking cramped.

````python
# Create a 2 x 2 grid, but make the whole figure larger
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

axes[0, 0].plot([1, 2, 3], [2, 4, 6])
axes[0, 0].set_title("Plot 1")

axes[0, 1].plot([1, 2, 3], [3, 6, 9])
axes[0, 1].set_title("Plot 2")

axes[1, 0].plot([1, 2, 3], [4, 8, 12])
axes[1, 0].set_title("Plot 3")

axes[1, 1].plot([1, 2, 3], [5, 10, 15])
axes[1, 1].set_title("Plot 4")

plt.tight_layout()
plt.show()
````

## Sharing Axes with `sharex` and `sharey`

When subplots show related data on the same scale, sharing the x-axis or y-axis avoids
repeating axis labels and makes the plots easier to compare side by side.

````python
# sharey=True: all subplots use the same y-axis scale
fig, axes = plt.subplots(1, 3, figsize=(10, 3), sharey=True)

axes[0].plot([1, 2, 3], [2, 4, 6])
axes[0].set_title("Plot 1")

axes[1].plot([1, 2, 3], [1, 3, 5])
axes[1].set_title("Plot 2")

axes[2].plot([1, 2, 3], [3, 5, 9])
axes[2].set_title("Plot 3")

# Only the first subplot needs a y-axis label now — they all share the same scale
axes[0].set_ylabel("Value")

plt.tight_layout()
plt.show()
````

## Giving the Whole Figure a Title with `suptitle`

Each subplot can have its own `set_title()`, but the FIGURE itself can also have one
overall title using `fig.suptitle()` — useful when all the subplots belong to one story.

````python
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

axes[0, 0].plot([1, 2, 3], [10, 20, 30])
axes[0, 0].set_title("Marks")

axes[0, 1].plot([1, 2, 3], [100, 200, 300])
axes[0, 1].set_title("Sales")

axes[1, 0].plot([1, 2, 3], [25, 30, 35])
axes[1, 0].set_title("Temperature")

axes[1, 1].plot([1, 2, 3], [50, 60, 70])
axes[1, 1].set_title("Population")

# One title for the entire figure, above all four subplots
fig.suptitle("Quarterly Dashboard", fontsize=16)

plt.tight_layout()
plt.show()
````

## Mixing Different Plot Types in One Grid

Subplots don't all have to be line plots. Each Axes is independent, so one grid can
combine a line plot, a bar chart, a scatter plot, and a histogram together.

````python
import numpy as np

fig, axes = plt.subplots(2, 2, figsize=(8, 6))

# Line plot
axes[0, 0].plot([1, 2, 3, 4], [10, 20, 15, 25])
axes[0, 0].set_title("Line Plot")

# Bar chart
axes[0, 1].bar(["A", "B", "C"], [5, 8, 3])
axes[0, 1].set_title("Bar Chart")

# Scatter plot
axes[1, 0].scatter([1, 2, 3, 4], [4, 1, 3, 2])
axes[1, 0].set_title("Scatter Plot")

# Histogram
data = np.random.randn(200)
axes[1, 1].hist(data, bins=15)
axes[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
````

## Manual Spacing with `subplots_adjust`

`plt.tight_layout()` adjusts spacing automatically, but sometimes we want exact control.
`plt.subplots_adjust()` sets the spacing ourselves using `hspace` (height) and `wspace` (width).

````python
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

axes[0, 0].plot([1, 2, 3], [2, 4, 6]); axes[0, 0].set_title("Plot 1")
axes[0, 1].plot([1, 2, 3], [3, 6, 9]); axes[0, 1].set_title("Plot 2")
axes[1, 0].plot([1, 2, 3], [4, 8, 12]); axes[1, 0].set_title("Plot 3")
axes[1, 1].plot([1, 2, 3], [5, 10, 15]); axes[1, 1].set_title("Plot 4")

# hspace = space between rows, wspace = space between columns
plt.subplots_adjust(hspace=0.5, wspace=0.4)

plt.show()
````

## Uneven Layouts with `GridSpec`

`plt.subplots()` always gives an evenly sized grid. When we need one plot to be wider or
taller than the others — like one big plot on top and two small ones below — `GridSpec`
lets us design the layout ourselves.

````python
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize=(8, 6))
gs = gridspec.GridSpec(2, 2, figure=fig)

# Top row: one wide plot spanning both columns
ax1 = fig.add_subplot(gs[0, :])
ax1.plot([1, 2, 3, 4], [10, 20, 15, 25])
ax1.set_title("Wide Plot (spans both columns)")

# Bottom row: two smaller plots, one per column
ax2 = fig.add_subplot(gs[1, 0])
ax2.bar(["A", "B", "C"], [5, 8, 3])
ax2.set_title("Bottom Left")

ax3 = fig.add_subplot(gs[1, 1])
ax3.scatter([1, 2, 3], [3, 1, 2])
ax3.set_title("Bottom Right")

plt.tight_layout()
plt.show()
````

## Saving a Multi-Subplot Figure

`fig.savefig()` saves the WHOLE figure — every subplot in it — to a file. `dpi` controls
the resolution, and `bbox_inches="tight"` trims extra whitespace around the edges.

````python
fig, axes = plt.subplots(1, 2, figsize=(8, 3))

axes[0].plot([1, 2, 3], [10, 20, 30])
axes[0].set_title("Student Marks")

axes[1].bar(["Jan", "Feb", "Mar"], [100, 200, 150])
axes[1].set_title("Sales")

plt.tight_layout()
fig.savefig("dashboard.png", dpi=150, bbox_inches="tight")
plt.show()

print("Saved dashboard.png")
````

## Quick Reference: Subplot Tools

| Function | What it does |
|---|---|
| `plt.subplot(rows, cols, pos)` | Selects one plot at a time in a grid |
| `plt.subplots(rows, cols)` | Creates the figure and all Axes together |
| `figsize=(w, h)` | Sets the overall figure size in inches |
| `sharex=True` / `sharey=True` | Shares an axis scale across subplots |
| `fig.suptitle("...")` | Adds one title above the whole figure |
| `plt.tight_layout()` | Adjusts spacing automatically |
| `plt.subplots_adjust(hspace, wspace)` | Adjusts spacing manually |
| `GridSpec` | Builds grids with differently sized plots |
| `fig.savefig("name.png")` | Saves the entire figure, all subplots included |

````python
import matplotlib.pyplot as plt

# -------------------------
# Subplot 1 - Bar Chart
# -------------------------
plt.subplot(2, 2, 1)

fruits = ['Apple', 'Banana', 'Mango', 'Grapes']
sales = [50, 30, 45, 20]

plt.bar(fruits, sales, color='purple')
plt.title("Bar Chart")
plt.xlabel("Fruits")
plt.ylabel("Sales")


# -------------------------
# Subplot 2 - Line Chart
# -------------------------
plt.subplot(2, 2, 2)

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
temperature = [20, 22, 25, 23, 21]

plt.plot(days, temperature, marker='o')
plt.title("Line Chart")
plt.xlabel("Days")
plt.ylabel("Temperature")


# -------------------------
# Subplot 3 - Pie Chart
# -------------------------
plt.subplot(2, 2, 3)

labels = ['Sleep', 'Work', 'Leisure', 'Other']
hours = [8, 8, 4, 4]

plt.pie(hours, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")


# -------------------------
# Subplot 4 - Histogram
# -------------------------
plt.subplot(2, 2, 4)

scores = [55, 60, 65, 70, 72, 75, 75, 78,
          50, 62, 75, 80, 88, 90, 92, 95, 98]

plt.hist(scores, bins=5, color='purple', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Scores")
plt.ylabel("Frequency")


# Automatically adjust spacing
plt.tight_layout()

# Display all plots
plt.show()
````
